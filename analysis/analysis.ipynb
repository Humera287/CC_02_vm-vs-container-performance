```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# CPU Benchmark Data
threads = ['1 Thread', '2 Thread', '4 Thread', '8 Thread']

vm_cpu = [515.84, 883.55, 928.17, 905.17]
docker_cpu = [517.19, 894.38, 900.45, 914.42]

x = np.arange(len(threads))
width = 0.35

plt.figure(figsize=(12, 7))

plt.bar(
    x - width/2,
    vm_cpu,
    width,
    label='Virtual Machine',
    color='#9ed6d9',
    alpha=0.9
)

plt.bar(
    x + width/2,
    docker_cpu,
    width,
    label='Docker Container',
    color='#c4a7d6',
    alpha=0.9
)

plt.title('CPU Scalability Comparison', fontsize=16)
plt.xlabel('Threads')
plt.ylabel('Events per Second')
plt.xticks(x, threads)
plt.legend()
plt.grid(axis='y', alpha=0.25)

plt.tight_layout()

plt.savefig('cpu_scalability.png', dpi=300)
plt.show()
```

```python
# Overall normalized performance
# Virtual Machine is used as the baseline = 100%

metrics = [
    'CPU 1T',
    'CPU 2T',
    'CPU 4T',
    'CPU 8T',
    'Memory 1T',
    'Memory 2T',
    'Seq Read',
    'Seq Write',
    'Rand Read',
    'Rand Write',
    'Net Sender',
    'Net Receiver',
    'API Health',
    'API Compute',
    'API Memory'
]

vm_values = [
    100, 100, 100, 100,
    100, 100,
    100, 100, 100, 100,
    100, 100,
    100, 100, 100
]

docker_values = [
    (517.19 / 515.84) * 100,
    (894.38 / 883.55) * 100,
    (900.45 / 928.17) * 100,
    (914.42 / 905.17) * 100,

    (5152.43 / 9541.97) * 100,
    (6970.16 / 9880.38) * 100,

    (500 / 461) * 100,
    (291 / 358) * 100,
    (1767 / 1313) * 100,
    (1346 / 1331) * 100,

    (13.7 / 14.1) * 100,
    (10.3 / 14.1) * 100,

    (371.07 / 419.79) * 100,
    (10.76 / 12.24) * 100,
    (14.40 / 16.43) * 100
]

x = np.arange(len(metrics))
width = 0.35

plt.figure(figsize=(16, 7))

plt.bar(
    x - width/2,
    vm_values,
    width,
    label='Virtual Machine (baseline = 100)',
    color='#9ed6d9',
    alpha=0.9
)

plt.bar(
    x + width/2,
    docker_values,
    width,
    label='Docker Container',
    color='#c4a7d6',
    alpha=0.9
)

# Baseline line
plt.axhline(
    y=100,
    color='blue',
    linewidth=1
)

plt.title('Overall VM vs Docker Performance (Normalized)', fontsize=16)
plt.xlabel('Metric')
plt.ylabel('Relative Performance (% of VM)')

plt.xticks(
    x,
    metrics,
    rotation=45,
    ha='right'
)

plt.legend()
plt.grid(axis='y', alpha=0.25)

plt.tight_layout()

plt.savefig('overall_performance_dashboard.png', dpi=300)
plt.show()
```
